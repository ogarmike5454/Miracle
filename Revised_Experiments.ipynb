# Revised experiments for the ICRITO 2027 submission

This notebook runs the revised analysis and shows the main tables. Every number in the paper comes from the files written to `results/`.

```python
import subprocess, sys, json
import pandas as pd
subprocess.run([sys.executable, '-I', 'scripts/run_experiments.py', 'data/all_repos_combined.csv', 'results'], check=True)
subprocess.run([sys.executable, '-I', 'scripts/make_figures.py', 'results', 'figs'], check=True)
```

## Shared holdout (pooled and per repository)

```python
h = pd.read_csv('results/holdout_metrics.csv')
h[h.variant=='combined'][['model','subset','accuracy','macro_f1','mcc','auc','tn','fp','fn','tp']].round(3)
```

## Repeated 5×5 cross-validation by text variant

```python
cv = pd.read_csv('results/cv_pooled_folds.csv')
cv.groupby(['variant','model'])[['macro_f1','macro_f1_kubernetes','macro_f1_eclipse']].agg(['mean','std']).round(3)
```

## Within-repository references

```python
wr = pd.read_csv('results/cv_within_repo_folds.csv')
wr.groupby(['repo','variant','model'])[['macro_f1','auc']].agg(['mean','std']).round(3)
```

## Directional transfer

```python
t = pd.read_csv('results/transfer_metrics.csv')
t[['source','target','variant','model','threshold','macro_f1','auc','pred_bug_share','tn','fp','fn','tp']].round(3)
```
